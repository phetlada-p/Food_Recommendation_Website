```python
import re
def clean_ingredients(text):
    if pd.isna(text):
        return np.nan
    ingredients = text.split('|')
    ingredients = [
        re.sub(r'\s+', ' ', item).strip().lower()
        for item in ingredients
    ]
    ingredients = [item for item in ingredients if item]
    return ' | '.join(ingredients)

df['ingredients_en_clean'] = df['ingredients_en'].apply(
    clean_ingredients
)
print(df[['ingredients_en', 'ingredients_en_clean']].head(10))
```

```python
# Ingredient mapping dictionary
ingredient_mapping = {
    'chillies': 'chili',
    'chilli': 'chili',
    'chilies': 'chili',
    'eggs': 'egg',
    'vegetables': 'vegetable',
    'herbs': 'herb'
}

# Apply ingredient mapping
def standardize_ingredients(text):
    if pd.isna(text):
        return np.nan

    ingredients = text.split('|')

    standardized = [
        ingredient_mapping.get(item.strip(), item.strip())
        for item in ingredients
    ]

    return ' | '.join(standardized)

df['ingredients_en_standardized'] = (
    df['ingredients_en_clean'].apply(standardize_ingredients)
)

print(df[['ingredients_en_clean',
          'ingredients_en_standardized']].head(10))
```

```python
#Multi Hot Encoder
from sklearn.preprocessing import MultiLabelBinarizer

# Initialize encoder
mlb = MultiLabelBinarizer()

# Transform ingredient lists into binary matrix
ingredient_matrix = mlb.fit_transform(model_df['ingredients_list'])

# Create DataFrame
ingredient_features = pd.DataFrame(
    ingredient_matrix,
    columns=mlb.classes_,
    index=model_df.index
)

print("Ingredient Matrix Shape:", ingredient_features.shape)

print("\nIngredient Features:")
print(ingredient_features.head())
```

```python
def recommend_by_ingredients(user_ingredients, top_n=5):

    # Standardize user ingredients
    user_ingredients = [
        item.strip().lower()
        for item in user_ingredients
    ]

    # Get all available ingredients
    available_ingredients = set(mlb.classes_)

    # Check unknown ingredients
    unknown = [
        item for item in user_ingredients
        if item not in available_ingredients
    ]

    if unknown:
        print("Unknown ingredients:", unknown)

    # Keep only known ingredients
    user_ingredients = [
        item for item in user_ingredients
        if item in available_ingredients
    ]

    if not user_ingredients:
        print("No valid ingredients found!")
        return

    # Calculate ingredient similarity
    user_vector = mlb.transform([user_ingredients])

    scores = cosine_similarity(
        user_vector,
        ingredient_features
    )[0]

    # Get top recommendations
    top_indices = scores.argsort()[::-1][:top_n]

    print("\nRecommended Foods:")
    print("-------------------------")

    for rank, idx in enumerate(top_indices, start=1):
        food = model_df.iloc[idx]['th_name']
        print(f"{rank}. {food} (Similarity: {scores[idx]:.2f})")
```

```python
import random
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity

n_runs = 5
all_results = []

for run in range(n_runs):

    random.seed(42 + run)
    run_results = []

    for pos in range(len(model_df)):

        row = model_df.iloc[pos]
        ingredients = row['ingredients_list']

        if len(ingredients) < 2:
            continue

        # สุ่มวัตถุดิบ 50%
        test_size = max(1, len(ingredients) // 2)

        test_ingredients = random.sample(
            ingredients, test_size
        )

        # แปลงวัตถุดิบเป็น Vector
        user_vector = mlb.transform([test_ingredients])

        # คำนวณ Similarity
        scores = cosine_similarity(
            user_vector, ingredient_features
        )[0]

        # เลือก Top 5
        top_indices = scores.argsort()[::-1][:5]

        run_results.append({
            'run': run + 1,
            'actual_food': row['th_name'],
            'test_ingredients': test_ingredients,
            'top1_hit': pos == top_indices[0],
            'top3_hit': pos in top_indices[:3],
            'top5_hit': pos in top_indices[:5]
        })

    all_results.extend(run_results)

# รวมผลการทดสอบ
results_df = pd.DataFrame(all_results)

# แสดงผลแต่ละรอบ
summary = results_df.groupby('run').agg(
    Top1=('top1_hit', 'mean'),
    Top3=('top3_hit', 'mean'),
    Top5=('top5_hit', 'mean')
) * 100

print("ผลการทดสอบแต่ละรอบ")
print(summary.round(2))

print("\nค่าเฉลี่ยทั้ง 5 รอบ")
print("Top 1:", round(results_df['top1_hit'].mean() * 100, 2), "%")
print("Top 3:", round(results_df['top3_hit'].mean() * 100, 2), "%")
print("Top 5:", round(results_df['top5_hit'].mean() * 100, 2), "%")
```

```python
# Function แนะนำเมนูอาหาร
def recommend_menu(user_ingredients, top_n=5):

    # 1. แปลงวัตถุดิบภาษาไทยเป็นภาษาอังกฤษ
    translated, not_found = translate_ingredients(
        user_ingredients
    )

    # ตรวจสอบวัตถุดิบที่ไม่พบ
    if not_found:
        print("ไม่พบวัตถุดิบ:", not_found)

    if len(translated) == 0:
        print("ไม่มีวัตถุดิบที่สามารถใช้แนะนำเมนูได้")
        return pd.DataFrame()

    # 2. แปลงเป็น Set
    user_set = set(translated)

    # 3. แปลงเป็น Vector
    user_vector = mlb.transform([translated])

    # 4. คำนวณ Cosine Similarity
    scores = cosine_similarity(
        user_vector, ingredient_features
    )[0]

    recommendations = []

    # 5. คำนวณคะแนนของแต่ละเมนู
    for pos in range(len(model_df)):

        row = model_df.iloc[pos]

        menu_ingredients = set(row['ingredients_list'])

        # วัตถุดิบที่มีร่วมกัน
        matched = user_set & menu_ingredients

        # ข้ามเมนูที่ไม่มีวัตถุดิบตรงกัน
        if len(matched) == 0:
            continue

        # วัตถุดิบที่ยังขาด
        missing = menu_ingredients - user_set

        # คำนวณ Coverage
        coverage = (
            len(matched) / len(menu_ingredients) * 100
            if menu_ingredients else 0
        )

        recommendations.append({
            'menu': row['th_name'],
            'similarity': scores[pos],
            'coverage': coverage,
            'matched': list(matched),
            'missing': list(missing),
            'status': (
                'ทำได้ทันที'
                if len(missing) == 0
                else 'ต้องซื้อเพิ่ม'
            )
        })

    # 6. แปลงผลลัพธ์เป็น DataFrame
    result = pd.DataFrame(recommendations)

    if result.empty:
        print("ไม่พบเมนูที่ตรงกับวัตถุดิบ")
        return result

    # 7. จัดลำดับเมนู
    result['_priority'] = result['status'].map({
        'ทำได้ทันที': 0,
        'ต้องซื้อเพิ่ม': 1
    })

    result = result.sort_values(
        by=['_priority', 'coverage', 'similarity'],
        ascending=[True, False, False]
    )

    return result.drop(
        columns=['_priority']
    ).head(top_n).reset_index(drop=True)
```

```python
#สร้างฟังก์ชันตรวจสอบผลลัพธ์
def validate_recommendations(results):

    if results.empty:
        print("ไม่มีเมนูให้ตรวจสอบ")
        return

    errors = []

    for index, row in results.iterrows():

        matched = set(row['matched'])
        missing = set(row['missing'])

        # 1. ตรวจสอบวัตถุดิบซ้ำกัน
        if matched & missing:
            errors.append(f"Row {index}: matched ซ้ำกับ missing")

        # 2. ตรวจสอบ Coverage
        total = len(matched) + len(missing)

        expected_coverage = (
            len(matched) / total * 100
            if total > 0 else 0
        )

        if abs(row['coverage'] - expected_coverage) > 0.01:
            errors.append(f"Row {index}: Coverage ไม่ถูกต้อง")

        # 3. ตรวจสอบ Status
        expected_status = (
            'ทำได้ทันที'
            if len(missing) == 0
            else 'ต้องซื้อเพิ่ม'
        )

        if row['status'] != expected_status:
            errors.append(f"Row {index}: Status ไม่ถูกต้อง")

    # 4. ตรวจสอบการเรียงลำดับ
    priority = {
        'ทำได้ทันที': 0,
        'ต้องซื้อเพิ่ม': 1
    }

    expected_order = results.assign(
        _priority=results['status'].map(priority)
    ).sort_values(
        by=['_priority', 'coverage', 'similarity'],
        ascending=[True, False, False]
    ).index.tolist()

    if results.index.tolist() != expected_order:
        errors.append("ลำดับเมนูไม่ถูกต้อง")

    # สรุปผล
    if errors:
        print("พบข้อผิดพลาด:")
        for error in errors:
            print("-", error)
    else:
        print("ผ่านการตรวจสอบทุกข้อ")
print("Test 1")
validate_recommendations(result_1)

print("\nTest 2")
validate_recommendations(result_2)

print("\nTest 3")
validate_recommendations(result_3)
```

`L-Top-K` นับเมนูที่ใช้วัตถุดิบชุดเดียวกันเป๊ะเป็นคำตอบที่ถูกด้วย (ใน dataset มีหลายเมนูที่วัตถุดิบเหมือนกันทุกอย่าง เช่น น้ำตาล+กะทิ+แป้ง)

```python
rows = []
for method, name in [("binary", "Binary (0/1 เดิม)"), ("tfidf", "TF-IDF")]:
    r = evaluate(method, n_runs=5)
    rows.append({
        "วิธี": name,
        "Top-1 (%)": round(r["strict"][1], 1),
        "Top-3 (%)": round(r["strict"][3], 1),
        "Top-5 (%)": round(r["strict"][5], 1),
        "L-Top-1 (%)": round(r["lenient"][1], 1),
        "L-Top-3 (%)": round(r["lenient"][3], 1),
        "L-Top-5 (%)": round(r["lenient"][5], 1),
    })

comparison = pd.DataFrame(rows)
print("จำนวนการทดสอบ:", r["tests"])
display(comparison)
```
